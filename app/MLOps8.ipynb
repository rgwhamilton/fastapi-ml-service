```python
from pathlib import Path
import textwrap
```

```python
# Creat Project Structure
folders = [
    "SentimentAPI/app",
    "SentimentAPI/tests"
]

for folder in folders:
    Path(folder).mkdir(parents=True, exist_ok=True)

print("Project structure created.")
```

```python
# Create FastAPI application

main_code = textwrap.dedent("""

from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class PredictionRequest(BaseModel):
    text: str

@app.get("/health")
def health():
    return {
        "status":"healthy"
    }

@app.post("/predict")
def predict(request:PredictionRequest):
    return {
        "prediction":"Positive",
        "confidence":0.98
    }
""")

Path("SentimentAPI/app/main.py").write_text(main_code)

print("main.py created")
```

```python
# Create pytest test file

test_code = textwrap.dedent("""

from fastapi.testclient import TestClient
from app.main import app

client = TestClient(app)

def test_health():
    response = client.get("/health")
    assert response.status_code == 200
    assert response.json()["status"] == "healthy"

def test_predict():
    response = client.post("/predict", json={"text":"This movie is excellent."})
    assert response.status_code == 200
    body = response.json()
    assert "prediction" in body
    assert "confidence" in body

def test_missing_text():
    response = client.post("/predict", json={})
    assert response.status_code == 422

def test_wrong_type():
    response = client.post("/predict", json={"text":123})
    assert response.status_code == 422

def test_empty_text():
    response = client.post("/predict", json={"text":""})
    assert response.status_code == 200

""")

Path("SentimentAPI/tests/test_api.py").write_text(test_code)

print("test_api.py created")
```

```python
# Run tests

print("""
Run the following commands
-----------------------------------------
cd SentimentAPI
pytest
-----------------------------------------
Expected Output
===============================
5 tests collected
5 passed
===============================
""")
```
